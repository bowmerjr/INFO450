Lesson 7: Iterating Loops

#For Loops

```python
fruits=["apple","orange","pear"]
```

```python
for fruit in fruits:
  print(fruit)
```

```python
for i in range(5):
  print(i)
```

```python
numbers=[10,20,30,40]
total=0
count=0
```

```python
for number in numbers:
  total=total+number
  count=count+1
  print(f"This is iteration {count} and the total is {total}.")
print(total)
```

```python
numbers=[3,5,7,2,8,1]
max_number=0
for number in numbers:
  if number>max_number:
    max_number=number
    print(number)
print(f"The biggest number is {max_number}.")
```

#Iterating over a string

```python
word="banana"
count=0
for letter in word:
  if letter=="a":
    count=count+1
print(f"There are {count} a's in the word banana")
```

```python
even_numbers=[]#empty list

for i in range(1,21):
  if i%2==0:
    even_numbers.append(i)
print(even_numbers)
```

#While Loops

```python
data=["yes","yes","no",1,"no",0,"yes"]
clean_data=[]

while len(data)>0:
  entry=data.pop(0)
  if entry !=0 and entry !=1:
    clean_data.append(entry)
clean_data
```

```python
sentence="The k-pop demon hunter uses music to fight demon."
words=sentence.split()
new_words=[]
for word in words:
  if word=="demon":
    new_words.append("monster")
  else:
      new_words.append(word)
```

```python
fruit_prices={"apple":1.2,"pear":2.1,"banana":2.6}
```

```python
for fruit in fruit_prices:
  print(fruit)
```

```python
for prices in fruit_prices.values():
  print(prices)
```

```python
for fruit,price in fruit_prices.items():
  print(fruit,price)
```
