# Logical Operators

```python
age=25
income=50000
#want to make sure they are an adult making 40000 a year
if age>18 and income>40000:
  print("Congrats on the loan")
```

```python
age=25
income=50000
#want to make sure they are an adult or making a lot of money
if (age>18 and income>40000) or income>4000000:
  print("Congrats on the loan")
```

```python
hours_worked=40
performance_score=85
tenure=2 #years with company
if (hours_worked>=40 and performance_score>80) or tenure>=2:
  print("Promoted")
else:
  print("Not")
```
